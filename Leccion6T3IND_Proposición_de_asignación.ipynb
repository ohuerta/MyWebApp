```python
#Ejemplo1L6T3IND Asignación Simple
# Configuración de tiempos de ciclo para 3 estaciones (en minutos)
estacion1, estacion2, estacion3 = 5, 8, 4#asignacion multiple de valores a las variables

# Cálculo del tiempo total de ciclo (Lead Time)
tiempo_total = estacion1 + estacion2 + estacion3#asignación de la suma a una variable
print(tiempo_total)  # Resultado: 17 minutos
```

```python
#Ejemplo2L6T3IND Asignación con Operadores Compuestos
# Control de inventario en almacén
inventario_acero = 500  # Toneladas iniciales

# Llega un camión con 150 toneladas más
inventario_acero += 150  # Equivale a: inventario_acero = inventario_acero + 150

# Se despachan 80 toneladas a producción
inventario_acero -= 80   # Equivale a: inventario_acero = inventario_acero - 80

print(inventario_acero)  # Resultado: 570 toneladas
```

```python
#Ejemplo3L6T3IND Intercambio de Valores (Swap)
# Prioridad de producción original
maquina_prioritaria = "Prensa 1"
maquina_secundaria = "Torno 2"

# Intercambio de asignación por mantenimiento en Prensa 1
maquina_prioritaria, maquina_secundaria = maquina_secundaria, maquina_prioritaria

print(maquina_prioritaria)  # Resultado: "Torno 2"
```
